```python
# functions doing the same thing as above (just easier)
def compute_rates(df):
    target = df[df['corrAns'] == 'space']
    hits = (target['trials.response_2.keys'] == 'space').sum()
    misses = (target['trials.response_2.keys'] != 'space').sum()
    H = hits / len(target)
    M = misses / len(target)

    
    non_target = df[df['corrAns'] != 'space']
    false_alarms = (non_target['trials.response_2.keys'] == 'space').sum()
    correct_rejections = (non_target['trials.response_2.keys'] != 'space').sum()
    F = false_alarms / len(non_target)
    R = correct_rejections / len(non_target)
    
    return H, M, F, R
    
H_active, M_active, F_active, R_active = compute_rates(active_trials)
H_sham,   M_sham,   F_sham,   R_sham   = compute_rates(sham_trials)
```

```python
def d_prime(hit_rate, false_alarm_rate):
    return norm.ppf(hit_rate) - norm.ppf(false_alarm_rate)

def criterion_c(hit_rate, false_alarm_rate):
    return -0.5 * (norm.ppf(hit_rate) + norm.ppf(false_alarm_rate))

def beta(hit_rate, false_alarm_rate):
    return np.exp((norm.ppf(false_alarm_rate)**2 - norm.ppf(hit_rate)**2) / 2)
```

```python
summary = pd.DataFrame({
    'Condition': ['Active', 'Sham'],
    'Hit Rate (H)': [H_active, H_sham],
    'False Alarm Rate (F)': [F_active, F_sham],
    'Miss Rate (M)': [M_active, M_sham],
    'Correct Rejection (R)': [R_active, R_sham],
    "d'": [d_prime(H_active, F_active), d_prime(H_sham, F_sham)],
    'Criterion c': [criterion_c(H_active, F_active), criterion_c(H_sham, F_sham)],
    'Bias β': [beta(H_active, F_active), beta(H_sham, F_sham)]
})

summary
```
